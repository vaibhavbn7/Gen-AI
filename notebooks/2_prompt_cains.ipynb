### Prompt templet - Dynamic Prompt

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser
import os
google_api_key = os.getenv("GOOGLE_API_KEY")
llm_google = ChatGoogleGenerativeAI(model="gemini-3.5-flash", google_api_key=google_api_key)
output = StrOutputParser()
```

```python
prompts = ChatPromptTemplate.from_messages([   
    {"role": "system", "content": "You are a Translator and translate input to {language}"},
    {"role": "user", "content": "{query}"}
    ])

prompts = prompts.format_messages(language="Kannada", query="Hello, how are you?")
res = llm_google.invoke(prompts)
print(res.content)
```

### Chains - Runnable

```python
prompts = ChatPromptTemplate.from_messages([
    ("system", "You are a Translator and translate input to {language}"),
    ("user", "{query}")
])

chains = prompts | llm_google | output

res = chains.invoke({
    "language": "Kannada",
    "query": "Hello, how are you?"
})

print(res)
```
