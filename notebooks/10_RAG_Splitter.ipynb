#### Document Loader

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("../data/speech.txt", encoding="utf-8")
texts = loader.load()
```

```python
pdfloader = PyPDFLoader("../data/medical_report.pdf")
pdf = pdfloader.load()
```

```python
# Split the text, pdf into chunks
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
text_chunks = splitter.split_documents(texts)
print(len(text_chunks))
pdf_chunks = splitter.split_documents(pdf)
print(len(pdf_chunks))
```
