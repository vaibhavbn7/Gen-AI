```python
from dotenv import load_dotenv
load_dotenv()
from langchain.tools import tool
```

```python
@tool
def add_numbers(num1: float, num2: float) -> float:
    """
        Add two numbers together.

        Args:
            num1 (float): The first number.
            num2 (float): The second number.
    """
    return num1 + num2

@tool
def multiply_numbers(num1: float, num2: float) -> float:
    """
        Multiply two numbers together.

        Args:
            num1 (float): The first number.
            num2 (float): The second number.
    """
    return num1 * num2
```

```python
add_numbers.invoke({"num1": 5, "num2": 10})
```

```python
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
```

```python
llm =  ChatOllama(model="gemma4:31b-cloud")
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="You are a helpful assistant that can perform basic arithmetic operations and always use tools for calculations."
)
```

```python
response = agent.invoke({"messages":[{"role": "user", "content": "What is the sum of 5 + 10 * 2?"}]})
```

```python
for res in response['messages']:
    print(res)
    print("\n")
```
