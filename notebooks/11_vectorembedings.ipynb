```python
from dotenv import load_dotenv
load_dotenv()
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
documents = [
    "Today, I want to speak from the heart about three pillars that define the future of our nation: our youth, our education system, and our farmers.",
    "The youth of our nation are the torchbearers of tomorrow. Their energy, creativity, and resilience are the driving forces that will shape our society. We must invest in their education, provide them with opportunities, and empower them to become leaders who will carry forward the legacy of our nation.",
    "Our education system is the bedrock upon which we build our future. It is imperative that we provide quality education to every child, regardless of their background. By fostering critical thinking, creativity, and a love for learning, we can equip our youth with the skills they need to thrive in an ever-changing world.",
    "Farmers are the backbone of our nation, ensuring that we have a steady supply of food and resources. Their hard work and dedication sustain our communities and fuel our economy. We must support our farmers by providing them with the necessary tools, technology, and policies that enable them to thrive in a competitive market.",]
```

```python
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview", output_dimensionality=1536)
```

```python
vector = embeddings.embed_documents(documents)
```

```python
print(len(vector[0]))
```

### Store Vector Embeddings in Vector DB

```python
from langchain_chroma import Chroma
```

```python
vector_store = Chroma.from_texts(
    texts=documents, 
    embedding=embeddings,
    persist_directory="./vector_db"
    )
```

```python
query = "What are the three pillars that define the future of our nation?"
results = vector_store.similarity_search(query=query, k=2)
results
```
